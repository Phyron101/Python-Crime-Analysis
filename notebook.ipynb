```python
# Re-run this cell
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

crimes = pd.read_csv("crimes.csv", dtype={"TIME OCC": str})
crimes["HOUR OCC"] = (crimes["TIME OCC"].str[:2].astype(int))
crimes["HOUR OCC"].head()  # .info removed, as .head() returns a DataFrame, not suitable for .info

sns.countplot(data=crimes, x="HOUR OCC")
plt.show()
peak_crime_hour = 12
night_time = crimes[crimes["HOUR OCC"].isin([22,23,0,1,2,3])] # this saves true or false values

# Fix: Remove 'rotation' from sns.countplot, set rotation on plt.xticks instead
sns.countplot(data=crimes, x="AREA NAME")
plt.xticks(rotation=90)
plt.show()
peak_night_crime_location = "Central"

age_bins = [0, 17, 25, 34, 44, 54, 64, np.inf]# they group age group by using bins and the pd.cut
age_labels = ["0-17", "18-25", "26-34", "35-44", "45-54", "55-64", "65+"]

# Add a new column using pd.cut() to bin values into discrete intervals
crimes["Age Bracket"] = pd.cut(crimes["Vict Age"],
                               bins=age_bins,
                               labels=age_labels)

# Find the category with the largest frequency
victim_ages = crimes["Age Bracket"].value_counts()
print(victim_ages)
```
